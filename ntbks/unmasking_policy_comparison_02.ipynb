```python
# Combine data into a DataFrame and melt for seaborn

df = pd.DataFrame(data)

df = df.melt(
    id_vars=["strategy", "nfe", "sampler"],
    value_vars=["exact_ppl", "gen_ppl", "entropy", "mauve"],
    var_name="metric_name",
    value_name="metric_value",
)

df.head()
```

# Plotting

```python
fig = sns.relplot(
    data=df,
    kind="line",
    x="nfe",
    y="metric_value",
    hue="sampler",
    col="metric_name",
    facet_kws={'sharey': False, 'sharex': True},
    marker="o",
    linewidth=2,
    markersize=10,
)

# Rename each of the column titles (arrow indicating up or down)
fig.axes[0, 0].set_title("DUEL Perplexity (↑)")
fig.axes[0, 1].set_title("Generative Perplexity (↑)")
fig.axes[0, 2].set_title("Entropy (↓)")
fig.axes[0, 3].set_title("MAUVE Score (↑)")

fig.axes[0, 0].set_yscale("log")
fig.axes[0, 1].set_yscale("log")
# fig.axes[0, 2].set_yscale("log")
fig.axes[0, 3].set_yscale("log")

# add grey dashed line labeled elbo at 23.52 in 0 subplot
elbo_value = 23.52
fig.axes[0, 0].axhline(y=elbo_value, color='gray', linestyle='--')
fig.axes[0, 0].text(
    x=fig.axes[0, 0].get_xlim()[1]*0.18,  # Slightly inset from right edge
    y=elbo_value*1.05,               # Slightly above the line
    s="ELBO",
    color='gray',
    fontsize=12,
    horizontalalignment='right'
)

# rename y axis label
fig.set_ylabels("Metric Value", fontsize=18)

# make ticks larger
for ax in fig.axes.flatten():
    ax.tick_params(axis='both', which='major', labelsize=14)
    


# fix y axis ticks to be more readable
fig.axes[0, 0].set_yticks([50, 100, 250])
fig.axes[0, 0].get_yaxis().set_major_formatter(plt.ScalarFormatter())
fig.axes[0, 1].set_yticks([50, 250, 1000])
fig.axes[0, 1].get_yaxis().set_major_formatter(plt.ScalarFormatter())
fig.axes[0, 3].set_yticks([0.01, 0.05, 0.3])
fig.axes[0, 3].get_yaxis().set_major_formatter(plt.ScalarFormatter())

# rename x and y axis labels
fig.set_xlabels("Num. of Function Evaluations", fontsize=18)
fig.set_ylabels("Metric Value", fontsize=18)

# Adjust legend to below plots and make legend horizontal
plt.subplots_adjust(bottom=0.35)
sns.move_legend(fig, "lower center", bbox_to_anchor=(0.5, 0.05), ncol=len(fig._legend.legend_handles), frameon=False, title="Unmasking Policy")

# make legend font larger
if fig._legend is not None:
    for text in fig._legend.get_texts():
        text.set_fontsize(18)
        
# make subfigure titles larger
for ax in fig.axes.flatten():
    ax.title.set_fontsize(20)   

# save as pdf high resolution
fig.savefig("unmasking_policy_comparison_02.pdf", dpi=300)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Set global scaling context for publication
# "paper" is the baseline; font_scale=2.0 provides significant enlargement
sns.set_context("paper", font_scale=1.5)
sns.set_style("whitegrid")

df_filtered = df[df['metric_name'].isin(['exact_ppl', 'gen_ppl'])]

fig = sns.relplot(
    data=df_filtered,
    kind="line",
    x="nfe",
    y="metric_value",
    hue="sampler",
    row="metric_name",
    height=5,
    aspect=1.3,
    facet_kws={'sharey': False, 'sharex': True},
    marker="o",
    linewidth=2.8,
    # make marker size larger
    markersize=12,
)

# 2. Legend Extraction and Customization
legend_data = fig._legend_data
handles = list(legend_data.values())
labels = list(legend_data.keys())
fig._legend.remove()

# Adding legend with explicit title/label sizes
ax_top = fig.axes[0, 0]
ax_top.legend(
    handles=handles, 
    labels=labels, 
    title="Unmasking Rule", 
    loc="upper right",
    frameon=True,
    fontsize='medium',      # Scales relative to font_scale
    title_fontsize='large'
)

# 3. Specific Font Size Overrides
# Titles for each facet
fig.axes[0, 0].set_title("DUEL Perplexity", fontsize=24, pad=20)
fig.axes[1, 0].set_title("Generative Perplexity", fontsize=24, pad=20)

# log scale for y axes
fig.axes[0, 0].set_yscale("log")
fig.axes[1, 0].set_yscale("log")

# fix y axis ticks to be more readable
fig.axes[0, 0].set_yticks([50, 100, 250])
fig.axes[0, 0].get_yaxis().set_major_formatter(plt.ScalarFormatter())
fig.axes[1, 0].set_yticks([50, 200, 1000])
fig.axes[1, 0].get_yaxis().set_major_formatter(plt.ScalarFormatter())

# Global Axis Labels
fig.set_xlabels("Number of Function Evaluations (NFE)", fontsize=22)
fig.set_ylabels("Perplexity", fontsize=22)

# Tick Parameters (Affects both subplots)
for ax in fig.axes.flat:
    ax.tick_params(axis='both', which='major', labelsize=18)
    
# add grey dashed line labeled elbo at 23.52 in 0 subplot
elbo_value = 23.52
fig.axes[0, 0].axhline(y=elbo_value, color='gray', linestyle='--', linewidth=2.5)
fig.axes[0, 0].text(
    x=fig.axes[0, 0].get_xlim()[1]*0.15,  # Slightly inset from right edge
    y=elbo_value*1.05,               # Slightly above the line
    s="ELBO",
    color='gray',
    fontsize=16,
    horizontalalignment='right'
)

plt.tight_layout()
plt.show()

# save with high dpi as pdf
fig.savefig("unmasking_policy_comparison.pdf", dpi=300)
```
