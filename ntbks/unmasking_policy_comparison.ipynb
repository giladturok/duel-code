# Plotting

```python
fig = sns.relplot(
    data=df,
    kind="line",
    x="nfe",
    y="metric_value",
    hue="strategy",
    col="metric_name",
    facet_kws={'sharey': False, 'sharex': True},
    marker="o"
)

fig.axes[0, 0].set_yscale("log")
fig.axes[0, 1].set_yscale("log")
fig.axes[0, 3].set_yscale("log")
```
