# Analysis

```python
# build unmelted dataframe from the data dict of dictss
data_frames = []
for dataset_name, methods_dict in data.items():
    for method_name, metrics_dict in methods_dict.items():
        filtered_metrics_dict = {k: v for k, v in metrics_dict.items() if v != "N/A"}
        row = {"dataset": dataset_name, "method": method_name}
        row.update(filtered_metrics_dict)
        data_frames.append(pd.DataFrame([row]))
res_df = pd.concat(data_frames, ignore_index=True)
res_df.head()
```

```python
melted_df = res_df.melt(
    id_vars=["dataset", "method"], 
    var_name="metric", 
    value_name="value",
)

# filter by acc_norm only
melted_df = melted_df[melted_df["metric"].isin(["ece,none"])]

fig = sns.catplot(
    data=melted_df,
    kind="bar",
    col="metric",
    col_wrap=3,
    x="dataset",
    y="value",
    sharey=False,
    hue="method",
    palette="Set2",
    aspect=2,
)

# Iterate through each subplot (axis)
for ax in fig.axes.flat:
    for container in ax.containers:
        ax.bar_label(container, fmt='%.2f', padding=3)
        
# move legend to inside the plot area
plt.legend(loc='upper left', bbox_to_anchor=(1, 1), fontsize='small')
plt.show()
```
